```python
#Step 1: Enter the data
#INPUT DATA :STUDY HOURS
X=[[1],[2],[3],[4],[5]]
#OUTPUT DATA:STUDENT MARKS
Y=[35,45,55,65,75]
print("Study Hours:",X)
print("Student Marks:",Y)
```

```python
#step2:Import Linear Regression
from sklearn.linear_model import LinearRegression
#here sklearn->scikit-learn is library used by Ml
#linear model contains the linear regression models
#Linear Regression is used to relationship between input and output
#from and import brig class into our python programes
```

```python
#Step 3: Create the model
model=LinearRegression()
#model->it will stores Ml model
#linear regression create a new linear regression model
```

```python
#Step 4: Train the model
model.fit(X,Y)
#fit->it will trains the model based on supplied data
#X,Y->input,output
```

```python
#Step 5: Predict marks for 6 study hours
prediction=model.predict([[6]])
#prediction=model.predict([[8]])
print("Predicted Marks:",prediction[0])
```

```python
print("Slope:",model.coef_[0])
print("Intercept:",model.intercept_)
```

```python
#step 6
import matplotlib.pyplot as plt
plt.scatter(X,Y,label="Actual Marks")
plt.plot(X,model.predict(X),label="Prediction Line")
plt.xlabel("Study Hours")
plt.ylabel("Student Marks")
plt.title("Student Marks Prediction")
plt.legend()
plt.grid(True)
plt.show()
```
